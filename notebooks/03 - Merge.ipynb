```python
import numpy as np
import re
```

```python
def _to_number(token: str) -> float | None:
    """Convert '$130K' / '190K' / '28' / '$120,000' into a float."""
    if token is None:
        return None
    s = str(token).strip().lower()
    if s in {"", "nan", "none"}:
        return None

    s = s.replace("$", "").replace(",", "").replace("+", "").strip()

    mult = 1.0
    if s.endswith("k"):
        mult = 1_000.0
        s = s[:-1].strip()
    elif s.endswith("m"):
        mult = 1_000_000.0
        s = s[:-1].strip()

    try:
        return float(s) * mult
    except ValueError:
        return None
```

```python
_num_pat = re.compile(r"\$?\s*\d+(?:\.\d+)?\s*[km]?", re.IGNORECASE)
```

```python
def split_salary_and_avg(salary_text: str):
    """
    Returns (salary_year, salary_hourly, salary_year_avg, salary_hour_avg)

    - salary_year: original string if unit is yearly or weekly (so you can audit)
    - salary_hourly: original string if unit is hourly
    - salary_year_avg:
        * if /yr -> avg as-is
        * if /wk -> avg * 52 (annualized)
    - salary_hour_avg:
        * if /hr -> avg as-is
    """
    if salary_text is None or (isinstance(salary_text, float) and np.isnan(salary_text)):
        return (pd.NA, pd.NA, pd.NA, pd.NA)

    txt = str(salary_text).strip()
    if txt == "" or txt.lower() == "nan":
        return (pd.NA, pd.NA, pd.NA, pd.NA)

    low = txt.lower()

    # Unit detection (more forgiving)
    is_hr = ("/hr" in low) or ("/ hr" in low) or ("per hour" in low) or ("hourly" in low)
    is_yr = ("/yr" in low) or ("/ yr" in low) or ("per year" in low) or ("yearly" in low) or ("annual" in low)
    is_wk = ("/wk" in low) or ("/ wk" in low) or ("per week" in low) or ("weekly" in low)

    # Extract 1–2 numbers (range or single value)
    tokens = _num_pat.findall(low)
    nums = [_to_number(t) for t in tokens]
    nums = [n for n in nums if n is not None]

    if not nums:
        return (pd.NA, pd.NA, pd.NA, pd.NA)

    if len(nums) == 1:
        avg = nums[0]
    else:
        a, b = nums[0], nums[1]
        if a > b:
            a, b = b, a
        avg = (a + b) / 2.0

    # Assign by unit
    if is_hr and not is_yr and not is_wk:
        return (pd.NA, txt, pd.NA, avg)

    if is_yr and not is_hr and not is_wk:
        return (txt, pd.NA, avg, pd.NA)

    if is_wk and not is_hr and not is_yr:
        # annualize weekly to yearly avg
        return (txt, pd.NA, avg * 52.0, pd.NA)

    # If multiple/ambiguous units appear, fall back to a reasonable heuristic
    # (you can tweak these cutoffs if needed)
    if max(nums) <= 500:
        # likely hourly
        return (pd.NA, txt, pd.NA, avg)
    else:
        # likely yearly-ish; if it smells like weekly, annualize
        if is_wk:
            return (txt, pd.NA, avg * 52.0, pd.NA)
        return (txt, pd.NA, avg, pd.NA)
```

```python
salary_col = "salary_rate"
```

```python
out = df2[salary_col].apply(split_salary_and_avg)
df2[["salary_year", "salary_hourly", "salary_year_avg", "salary_hour_avg"]] = pd.DataFrame(
    out.tolist(), index=df2.index
)
```

```python
df2["salary_year_avg"] = pd.to_numeric(df2["salary_year_avg"], errors="coerce")
df2["salary_hour_avg"] = pd.to_numeric(df2["salary_hour_avg"], errors="coerce")
```
